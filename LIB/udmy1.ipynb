## 1. Importación y carga de los datos

En esta sección se importa la librería **Pandas**, la cual permite trabajar con datos organizados en tablas.

Posteriormente, se define la URL donde se encuentra el conjunto de datos de casos de COVID-19 de la ciudad de Nueva York. Mediante la función `pd.read_csv()` se obtiene la información y se almacena en el DataFrame llamado `covid`.

Finalmente, se utiliza `covid.head()` para visualizar las primeras filas del conjunto de datos y comprobar que la información se cargó correctamente.

**Funciones utilizadas:**
- `import pandas as pd`: importa la librería Pandas.
- `pd.read_csv(url)`: carga los datos desde la URL.
- `covid.head()`: muestra las primeras 5 filas del DataFrame.

```python
import pandas as pd
url = "https://data.cityofnewyork.us/resource/rc75-m7u3.csv"
#covid = pd.read_csv(url)
covid = pd.read_csv(url)
covid.head()
```

```python
covid.info()
```

## 2. Exportación del conjunto de datos

En esta sección se utiliza la función `to_csv()` de Pandas para guardar el DataFrame `covid` en un archivo con formato **CSV**.

El archivo se guarda con el nombre **`NYC_covid_cases.csv`**, lo que permite conservar los datos obtenidos y utilizarlos posteriormente sin necesidad de volver a descargarlos desde la URL.

**Función utilizada:**
- `covid.to_csv('NYC_covid_cases.csv')`: exporta el contenido del DataFrame `covid` a un archivo CSV.



```python
covid.to_csv('NYC_covid_cases.csv')
```

## 3. Selección y exportación de variables específicas

En esta sección se utiliza la función `to_csv()` para crear un nuevo archivo CSV a partir del DataFrame `covid`. A diferencia del archivo anterior, en este caso se seleccionan únicamente tres columnas de interés: **número de casos**, **personas hospitalizadas** y **fallecimientos** por COVID-19.

El nuevo archivo se guarda con el nombre **`NYC1_covid_cases.csv`**.

**Columnas seleccionadas:**
- `case_count`: cantidad de casos de COVID-19.
- `hospitalized_count`: cantidad de personas hospitalizadas.
- `death_count`: cantidad de fallecimientos.

**Función utilizada:**
- `covid.to_csv()`: permite exportar los datos seleccionados a un archivo CSV.
- `columns = [...]`: indica las columnas específicas que se desean guardar.



```python
covid.to_csv('NYC1_covid_cases.csv', columns = ['case_count', 'hospitalized_count', 'death_count'])
```
