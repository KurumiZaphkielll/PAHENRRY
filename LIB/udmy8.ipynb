# Eliminación de columnas de un DataFrame con Pandas

En este programa se carga un archivo de Excel con información de la NBA y se muestra cómo eliminar una o varias columnas del DataFrame utilizando `drop()`.

## 1. Importación, carga y visualización de los datos

Se importa la librería **Pandas** y se utiliza `pd.read_excel()` para cargar el archivo `Nba Teams.xlsx` en el DataFrame `nba`.

Finalmente, `nba.head()` permite visualizar las primeras cinco filas para comprobar que los datos fueron cargados correctamente.

```python
import pandas as pd
nba = pd.read_excel("Nba Teams.xlsx")
nba.head()
```

## 2. Eliminación temporal de una columna

La instrucción `nba.drop("Salary", axis=1)` elimina temporalmente la columna `Salary` del DataFrame y muestra el resultado sin modificar el DataFrame original, ya que no se utiliza `inplace=True`.

```python
nba.drop("Salary", axis = 1)
```

## 3. Visualización del DataFrame original

La instrucción `nba` permite visualizar nuevamente el DataFrame original. Como la eliminación anterior no utilizó `inplace=True`, la columna `Salary` permanece en el DataFrame.

```python
nba
```

## 4. Eliminación permanente de una columna

La instrucción `nba.drop("Salary", axis=1, inplace=True)` elimina la columna `Salary` del DataFrame `nba` de forma permanente.

El parámetro `axis=1` indica que se está eliminando una columna, mientras que `inplace=True` aplica directamente el cambio sobre el DataFrame original.

```python
nba.drop("Salary", axis = 1, inplace = True)
```

## 5. Visualización del DataFrame actualizado

La instrucción `nba` permite visualizar nuevamente el DataFrame después de eliminar permanentemente la columna `Salary`, comprobando los cambios realizados en su estructura.

```python
nba
```

## 6. Eliminación de varias columnas

La instrucción `nba.drop(["Team", "Division"], axis=1, inplace=True)` elimina las columnas `Team` y `Division` del DataFrame `nba` de forma permanente.

El uso de una lista permite eliminar varias columnas al mismo tiempo. El parámetro `axis=1` indica que se están eliminando columnas y `inplace=True` aplica los cambios directamente sobre el DataFrame original.

```python
nba.drop(["Team", "Division"], axis = 1, inplace = True)
```

## 7. Visualización del DataFrame final

La instrucción `nba` permite visualizar el DataFrame después de eliminar las columnas `Team` y `Division`, mostrando la estructura final de los datos.

```python
nba
```
