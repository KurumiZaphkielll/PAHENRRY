# Optimización y manipulación de datos de empleados con Pandas

En este programa se carga y explora un conjunto de datos de empleados. Se analiza la información del DataFrame, se cuentan los valores únicos, se convierten las columnas `department` y `position` al tipo `category` y se realizan operaciones de texto para modificar el formato de los nombres.

## 1. Importación y carga de los datos

Se importa la librería **Pandas** y se carga el archivo `Employee.csv` desde Google Drive. Los datos se almacenan en el DataFrame `emp`.

Finalmente, `emp.head()` permite visualizar las primeras cinco filas del conjunto de datos para comprobar que la información fue cargada correctamente.

```python
import pandas as pd
emp = pd.read_csv("/content/drive/MyDrive/udmy/dataAnalisys/Employee.csv")
emp.head()
```

## 2. Información del DataFrame

La instrucción `emp.info()` muestra información general del DataFrame `emp`, incluyendo el número de filas y columnas, los nombres de las columnas, la cantidad de valores no nulos y los tipos de datos de cada columna.

Esto permite conocer la estructura del conjunto de datos.

```python
emp.info()
```

## 3. Conteo de valores únicos

La instrucción `emp.nunique()` cuenta la cantidad de valores diferentes que existen en cada columna del DataFrame `emp`.

Esto permite identificar cuántos valores únicos tiene cada variable del conjunto de datos.

```python
emp.nunique()
```

## 4. Conversión de columnas a categoría

La instrucción `emp[["department", "position"]] = emp[["department", "position"]].astype("category")` convierte las columnas `department` y `position` al tipo de dato `category`.

Esto permite representar de forma más eficiente columnas que contienen un conjunto limitado de categorías repetidas, como departamentos y cargos.



```python
emp[["department", "position"]] = emp[["department", "position"]].astype("category")
```

## 5. Verificación de los tipos de datos

La instrucción `emp.info()` muestra nuevamente la información del DataFrame `emp` para comprobar los cambios realizados.

En particular, permite verificar que las columnas `department` y `position` ahora tienen el tipo de dato `category`.

```python
emp.info()
```

## 6. Conversión de texto a mayúsculas y minúsculas

La instrucción `"happy".upper()` convierte el texto `"happy"` a mayúsculas, obteniendo `"HAPPY"`.

Por otra parte, `"PYTHON".lower()` convierte el texto `"PYTHON"` a minúsculas, obteniendo `"python"`.

```python
"happy".upper()
"PYTHON".lower()
```

## 7. Conversión del formato de los nombres

La instrucción `emp["name"]` permite seleccionar y visualizar la columna `name` del DataFrame `emp`.

Luego, `emp["name"].str.lower().str.title()` convierte primero todos los nombres a minúsculas mediante `lower()` y después aplica formato de título mediante `title()`, colocando la primera letra de cada palabra en mayúscula.

Esta operación solo muestra el resultado transformado y no modifica la columna original.

```python
emp["name"]
emp["name"].str.lower().str.title()
```
